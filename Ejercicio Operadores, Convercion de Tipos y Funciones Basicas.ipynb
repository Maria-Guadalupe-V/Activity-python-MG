```python
import ast

# Inicializar acumuladores
resultado_a = 0
resultado_b = 0
resultado_c = 0

# Funciones para operaciones
def res(iterable):
    items = list(iterable)
    result = items[0]
    for i in items[1:]:
        result -= i
    return result

def prod(iterable):
    result = 1
    for i in iterable:
        result *= i
    return result

def div(iterable):
    items = list(iterable)
    result = items[0]
    for i in items[1:]:
        result /= i
    return result

def pot(iterable):
    items = list(iterable)
    result = items[0]
    for i in items[1:]:
        result **= i
    return result

while True:
    print("\n===CALCULADORA===")
    print("Elige una opción:")
    print("1 - Suma")
    print("2 - Resta")
    print("3 - Multiplicación")
    print("4 - División")
    print("5 - Potencia")
    print("6 - Salir")

    x = input("Escribe tu elección: ")

    if x in ["1", "2", "3", "4", "5"]:
        for i in range(1, 6):
            print(i)
        if i > 6:
            print("El contador se ha pasado de 6")

        try:
            a = ast.literal_eval(input("Introduce una lista de 5 datos (ej: [1, 2, 3, 4, 5]):\n"))
            if not (isinstance(a, list) and len(a) == 5):
                print("Entrada inválida: debe ser una lista de 5 elementos.")
                continue

            b = ast.literal_eval(input("Introduce un set de 5 datos (ej: {1, 2, 3, 4, 5}):\n"))
            if not (isinstance(b, set) and len(b) == 5):
                print("Entrada inválida: debe ser un set de 5 elementos.")
                continue

            c = ast.literal_eval(input("Introduce un diccionario con 5 pares clave:valor (ej: {'a':1, 'b':2, 'c':3, 'd':4, 'e':5}):\n"))
            if not (isinstance(c, dict) and len(c) == 5):
                print("Entrada inválida: debe ser un diccionario con 5 pares.")
                continue

            if x == "1":
                r_a = sum(a)
                r_b = sum(b)
                r_c = sum(c.values())
                resultado_a += r_a
                resultado_b += r_b
                resultado_c += r_c
                op = "Suma"

            elif x == "2":
                r_a = res(a)
                r_b = res(b)
                r_c = res(c.values())
                resultado_a -= r_a
                resultado_b -= r_b
                resultado_c -= r_c
                op = "Resta"

            elif x == "3":
                r_a = prod(a)
                r_b = prod(b)
                r_c = prod(c.values())
                resultado_a *= r_a if resultado_a else r_a
                resultado_b *= r_b if resultado_b else r_b
                resultado_c *= r_c if resultado_c else r_c
                op = "Multiplicación"

            elif x == "4":
                r_a = div(a)
                r_b = div(b)
                r_c = div(c.values())
                resultado_a = r_a if resultado_a == 0 else resultado_a / r_a
                resultado_b = r_b if resultado_b == 0 else resultado_b / r_b
                resultado_c = r_c if resultado_c == 0 else resultado_c / r_c
                op = "División"

            elif x == "5":
                r_a = pot(a)
                r_b = pot(b)
                r_c = pot(c.values())
                resultado_a = r_a
                resultado_b = r_b
                resultado_c = r_c
                op = "Potencia"

            # Mostrar resultados
            print(f"{op} de la lista: {r_a}")
            print(f"{op} del set: {r_b}")
            print(f"{op} de los valores del diccionario C: {r_c}")

        except Exception as e:
            print("Error en la entrada:", e)

    elif x == "6":
        print("Saliendo de la calculadora...")
        break

    else:
        print("Opción aún no implementada o inválida.")
```
