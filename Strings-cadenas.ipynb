# Indexacion de Cadenas

- ¿Que es un indice?
- ¿Por que es necesario?

- Indexación significa acceder a los elementos
- Para acceder a los elemmentos podemos usar corchetes "[.]"
- El indice o número que pongamos se refiere a la posicion dentro de la lista ordenada
- La indexación puede usarse para extraer elementos individuales de una estructura de datos
- En Python, toda la indexación parte desde el numero 0 como primera posicion

- En Python, tenemos dos tipos de indices:
  - Positivos: comienzan desde 0; estos indices buscan de izquierda a derecha la posision solisitada
  - Negativos: comienzan desde -1; estos indices buscan de derecha a izquierda la posicion solicitada
          

# Segmentación de Cadenas

```python
# indices positivos 
my_string = "Curso de Backend Python"
print(my_string)
print(type(my_string))
print(my_string[0])
print(my_string[0:5]) # 0,1,2,3,4
print(my_string[:5]) # 0,1,2,3,4
print(my_string[:])  # todos 
print(my_string[6:11]) #  6,7,8,9,10
```

```python
# error
my_string = "Curso de Backend Python"
print(my_string[9999])
```

```python
# indices negativos
my_string = "Curso de Backend Python"
print(my_string[len(my_string)-1])
print(my_string[-1])
print(my_string[-5:])
```

```python
my_string_1 = "10"
my_string_2 = "20"
print(my_string_1+my_string_2)
print(eval(my_string_1+my_string_2))
print(eval("{}+{}".format(my_string_1, my_string_2)))
print(int(my_string_1)+int(my_string_2))
```

# Funciones de Cadenas

### Split
Divide la Cadena basandose en el separador especificado

```python
my_string = "Curso de Backend Python de EBAC te lleva de la mano para que aprendas a programar"

split_result = my_string.split(" ")

print(split_result)
```

```python
my_string = "use123@gmail.com"

username, domain = my_string.split("@")

print(username)
print(domain)
```

```python
my_string = """2022-09-28|error|Este es un mensaje de error
2022-09-27|warning|Este es un mensaje de advertencia
2022-09-25|info|Este es un mensaje informativo"""

logs = my_string.split("\n")
for log in logs: 
    date, log_level, message = log.split("|")
    print("date:", date)
    print("log_level:", log_level)
    print("message:", message)
```
