```python
import pokebase as pb
import requests

pokemon_to_search = input("Ingresa el nombre de un pokemon:\n").lower()

pokemon = pb.pokemon(pokemon_to_search)

id_ = pokemon.id
name = pokemon.name
height = pokemon.height
weight = pokemon.weight

print("ID:", id_)
print("Nombre:", name)
print("Altura:", height)
print("Peso:", weight)

url = f"https://pokeapi.co/api/v2/pokemon-species/{id_}/"
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    for entry in data["flavor_text_entries"]:
        if entry["language"]["name"] == "es":
            print("Descripción:", entry["flavor_text"])
            break
else:
    print("No se pudo obtener la descripción")
```
