```python
#Catalogo de laptops disponibles
laptopsEnElCatalogo = [
    {"Marca":"Dell","Precio": "$62,649","Modelo":"Dell Pro Max 16, 16","Ram":"32GB", "Procesador":"Intel Core Ultra 9 285H","Almacenamiento":"1TB","stock":"6 unidades"},
    {"Marca":"Lenovo","Precio":"$14,399","Modelo":"Lenovo V15", "Ram":"16GB", "Procesador":"Intel Core i7-1335U","Almacenamiento": "1TB", "stock": "10 unidades"},
    {"Marca":"hp","Precio":"$9,569","Modelo":"HP 240R G10 de 14", "Ram":"8GB", "Procesador":"Intel Core i5 120U","Almacenamiento": "512GB", "stock":"5 unidades"},
    {"Marca":"acer","Precio":"$4,499","Modelo":"Acer Aspire Lite 15", "Ram":"4GB", "Procesador":"Celeron N4500","Almacenamiento": "64GB", "stock":"16 unidades"},
    {"Marca":"asus","Precio":"$49,999","Modelo":"ASUS Zenbook A16 ", "Ram":"48GB", "Procesador":"Snapdragon X2 Elite Extreme","Almacenamiento": "1TB", "stock":"3 unidades"},
]
def pedirLaptop():# funcion que pide la laptop que el usuario quiere
    print("Que marca de laptop buscas")
    laptop = input("Ingresa la marca\n").lower()
    def buscarLaptop(laptop): # funcion que busca los modelos disponibles de la marca que el usuario dio
            bandera = False
            for lap in laptopsEnElCatalogo:
                if laptop == lap["Marca"].lower():
                    print("El modelo encontrado es")
                    for clave, valor in lap.items():
                        print( clave, ":", valor)
                    bandera = True
                    break
            if bandera == False: # condicion que se ejecuta para saber si el usuario quiere seguir buscando
                print("La laptop no esta disponible, desea buscar de nuevo?")
                respuesta = input("si o no: \n").lower()
                while respuesta != "si" and respuesta != "no":
                    print("Respuesta inválida, escribe 'si' o 'no'")
                    respuesta = input("si o no: \n").lower()
                if respuesta == "si":
                    pedirLaptop()
                elif respuesta == "no":
                    print("Adiós")
    buscarLaptop(laptop)
```
