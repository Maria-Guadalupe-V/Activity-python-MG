```python
my_tuple = (5, 4, 2, 7, 9 ,0, 1)
my_sorted_tuple = sorted(my_tuple)
my_sorted_tuple_2 = tuple(sorted(my_tuple))

print(type(my_tuple))
print(my_tuple)

print(type(my_sorted_tuple))
print(my_sorted_tuple)

print(type(my_sorted_tuple_2))
print(my_sorted_tuple_2)
```

# Convercion de Tuplas

##  De Cadena a Tupla 

```python
my_string = "Backend Python"
my_tuple = tuple(my_string)

print (my_string)
print (my_tuple )
```

## Lista a Tupla

```python
my_list = ["python", "javascript", "java", "go", "scala"]
my_tuple = tuple(my_list)

print(my_list)
print(my_tuple)
```

## Tupla a cadena

```python
my_tuple = ('B', 'a', 'c', 'k', 'e', 'n', 'd', ' ', 'P', 'y', 't', 'h', 'o', 'n')
my_string = "".join(my_tuple)


print(my_tuple)
print(my_string)
```

## Tupla a Lista

```python
my_tuple = ("python", "javascript", "java", "go", "scala")
my_list = list(my_list)

print(my_tuple)
print(my_list)
```

## Ventajas de la Tupla sobre la Lista

- Generalmente usamos Tuplas para elementos heterogéneos (de diferente tipo de dato), y usamos listas para elementos homogéneos (del mismo tipo de dato)
- Las tuplas son inmutables, iterar atravez de una tupla es mas rapido que con una lista. Por lo tanto hay una pequeña mejora de rendimiento
- Las tuplas que contienen elementos inmutables pueden der usadas como llaves para un diccionario. Con listas mo es posible
- Si tienes data que no cambia implementarla te garantiza que se mantendra protegida a modificaciones

```python
my_tuple[0] = "valor intruso"
```

```python
my_list[0] = "valor intruso"
```

```python
print(my_tuple)
print(my_list)
```
