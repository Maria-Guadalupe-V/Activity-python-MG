```python
while True:
    print("\nCarrito de compras\n")
    print("""
    Productos Disponibles:
    1.- Pan
    2.- Jamón
    3.- Crema
    4.- Mayonesa
    5.- Lechuga
    6.- Queso
    7.- Huevo
    8.- Cereal
    9.- Salir
    """)
    print("Ingresa una lista con los productos en este formato: [[Cantidad, Producto]]")
    print("Por ejemplo: [[1,1], [3,2]] significa 1 Pan y 3 Jamón")

    try:
        entrada = input("\nIngresa tu carrito: ")
        
        if entrada.lower() == "salir":
            print("Gracias por usar el programa. ¡Hasta luego!")
            break

        my_shopping_cart = eval(entrada)

        if isinstance(my_shopping_cart, list) and all(isinstance(item, list) and len(item) == 2 for item in my_shopping_cart):
            print("\nTu carrito es:")
            for cantidad, producto_id in my_shopping_cart:

                
                if producto_id == 1:
                    producto = "Pan"
                elif producto_id == 2:
                    producto = "Jamón"
                elif producto_id == 3:
                    producto = "Crema"
                elif producto_id == 4:
                    producto = "Mayonesa"
                elif producto_id == 5:
                    producto = "Lechuga"
                elif producto_id == 6:
                    producto = "Queso"
                elif producto_id == 7:
                    producto = "Huevo"
                elif producto_id == 8:
                    producto = "Cereal"
                else:
                    producto = "Producto desconocido"

                print(f"{cantidad} {producto}")
        else:
            print("Formato incorrecto, debes ingresar una lista de listas")
    except:
        print("Hubo un error con la entrada. Usa el formato correcto: [[Cantidad, Producto]] o escribe 'salir'")
```
