## Convercion  de Listas

### Cadenas a listas

```python
# separar palabras
my_string = "Curso de Backend Pyton"
my_list =  my_string.split()
print(my_string)
print(my_list)
```

```python
# separar caracteres
my_string = "Curso de Backend Pyton"
my_list =  list(my_string)
print(my_string)
print(my_list)
```

```python
# Separar caracteres en una lista de listas
my_string = "Curso de Backend Pyton"
my_list =  list(map(list, my_string ))
print(my_string)
print(my_list)
```

### Listas a Cadenas

```python
# usando espacios entre palabras
my_list = ['Curso', 'de', 'Backend', 'Pyton']
my_string = " ".join(my_list)
print(my_list)
print(my_string)
```

```python
# usando espacios entre palabras
my_list = ['C', 'u', 'r', 's', 'o', ' ', 'd', 'e', ' ', 'B', 'a', 'c', 'k', 'e', 'n', 'd', ' ', 'P', 'y', 't', 'o', 'n']

my_string = "".join(my_list)
print(my_list)
print(my_string)
```

```python
# usando espacios entre palabras
my_list = [['C'], ['u'], ['r'], ['s'], ['o'], [' '], ['d'], ['e'], [' '], ['B'], ['a'], ['c'], ['k'], ['e'], ['n'], ['d'], [' '], ['P'], ['y'], ['t'], ['o'], ['n']]
my_string = "".join(["".join(element) for element in my_list])
print(my_list)
print(my_string)
```

# Metodos para manipulacion de listas

## append

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.append("delfin")
print(my_list)
print(len(my_list))
```

```python
my_list.append("gato")
print(my_list)
print(len(my_list))
```

```python
my_list = [300, 500, 100, 200]
print(my_list)
print(len(my_list))
```

```python
my_list.append(111)
print(my_list)
print(len(my_list))
```

```python
my_list.append(333)
print(my_list)
print(len(my_list))
```

## insert

Agregar un valor a una lista en una pocision especifica

insert(índice,valor)

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.insert(2, "delfin")
print(my_list)
print(len(my_list))
```

## clear 

Elminar todos los elementos de la lista

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.clear()
print(my_list)
print(len(my_list))
```

```python
my_list = [300, 500, 100, 200]
print(my_list)
print(len(my_list))
```

```python
my_list.clear()
print(my_list)
print(len(my_list))
```

## copy 

Copia los elementos de una lista

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list_copy = my_list.copy()
print(my_list_copy)
print(len(my_list_copy))
```

```python
my_list_copy.clear()
print(my_list_copy)
print(len(my_list_copy))
print(my_list)
print(len(my_list))
```

```python
my_list_assigner = my_list
my_list_assigner.clear()
print(my_list_assigner)
print(len(my_list_assigner))
print(my_list)
print(len(my_list))
```

## pop

Elimina los elememtos de una lista basado en los indices

```python
my_list = [100,300,500,100,200,100,300]
print(my_list)
print(len(my_list))
```

```python
# elimina el indice -1 por default
my_list.pop()
print(my_list)
print(len(my_list))
```

```python
my_list.pop(2)
print(my_list)
print(len(my_list))
```
