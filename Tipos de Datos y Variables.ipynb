# Variables
### ¿Que es una variable
- Son un espacio temporal para guardar un valor con un nombre especifico
- No hay nesesidad de declarar el tipo de dato explicitamente
- Python es un lenguaje tipado dinamicamente, lo que quiere decir qu el tipo de dato se define mientras corre el programa
- Las variables en python funcionan diferente que en otros lenguajes de programacion
- No necesitamos crear las variables, solo asignamos un nombre a un valor usando el de asignacion

### Tipos de variables:
- Variables globales: aquellas que se utilizaran para todo el programa
- Variables locales: aquellas que se van autilizar solo en un bloque especifico

# Variables Globales
- Declaramos variables globales afuera de funciones
- El valor de la variable puede ser contante para todo el programa
- Cualquier funcion podra acceder a estas variables 
- Podemos decir que las variables globales tienen un alcance global
- Normalmente son mas usadas para constantes 
- Las variables globales mantienen sus valores a travez de la ejecucion completa de tu programa
- Una variable global puede ser accesada por cualquier funcion 

```python
# Vaiable Global 
my_global_variable=50

def my_funcion_1():
    pass

def my_funcion_2():
    pass
```

```python
# variables locales
def my_funcion_1():
    my_local_variable=5

def my_funcion_2():
     my_local_variable=35
```

```python
# Vaiable Global 
my_global_variable=50

def my_funcion_1():
    print("my_funcion_1 - El valor de my_global_variable es: ", my_global_variable)

def my_funcion_2():
     print("my_funcion_2 - El valor de my_global_variable es: ", my_global_variable)
```

# Variables locales
- Podemos usarlas dentro de una funcion
- Podemos tener una restriccion al acceder a una variable local desde otra funcion
- No podemos usarlas durante todo el pograma

- Lo que dure la funcion sera lo que dure el alcanse de nuestra variable

# Palabra reservada "global"

```python
def my_funcion_1():
    my_variable=10 # Local
    print("my_funcion_1 - El valor de my_variable es: ",my_variable)

def my_funcion_2():
    global  my_variable # global
    my_variable=20 # Local
    print("my_funcion_2 - El valor de my_variable es: ", my_variable)

def my_funcion_3():
    print("my_funcion_3 - El valor de my_variable es: ", my_variable)
```

```python
my_funcion_1()
```

```python
my_funcion_2()
```

```python
 my_funcion_3()
```

```python
def my_funcion_1():
    my_variable=10 # Local
    print("my_funcion_1 - El valor de my_variable es: ",my_variable)

def my_funcion_2():
    global a, b, c, d, e # global
    a=10
    b=20
    c=30
    d=40
    e=50
    print("my_funcion_2 - El valor de a es: ", a)
    print("my_funcion_2 - El valor de b es: ", b)
    print("my_funcion_2 - El valor de c es: ", c)
    print("my_funcion_2 - El valor de d es: ", d)
    print("my_funcion_2 - El valor de e es: ", e)

def my_funcion_3():
    print("my_funcion_2 - El valor de a es: ", a)
    print("my_funcion_2 - El valor de b es: ", b)
    print("my_funcion_2 - El valor de c es: ", c)
    print("my_funcion_2 - El valor de d es: ", d)
    print("my_funcion_2 - El valor de e es: ", e)
```

# Forma binaria
- codigo organizado en bytes
- Lennguaje Binario

### Valor vase: 2

- solamente acepta dos digitos (1 y 0)
- usamos el prefijo 'ob' o '0B'

```python
my_decimal_number = 11
print("El valor de my_decimal_number:  ",my_decimal_number)
```

```python
my_binary_number = 0b11 # 1*2**1 + 1*2**0
print("El valor de my_binary_number:  ",my_binary_number)
```

# Forma Octal

- Solamente acepta digitos del 0 al 7

```python
my_octal_number = 0o17 # 1*8**1 + 7*8**0
print("El valor de my_octal_number:  ",my_octal_number)
```

```python
my_decimal_number = 153
print("El valor de my_decimal_number:  ",my_decimal_number)
```

```python
my_octal_number = 0O153 # 1*8**2 + 5*8**1 + 3*8**0
print("El valor de my_octal_number:  ",my_octal_number)
```
