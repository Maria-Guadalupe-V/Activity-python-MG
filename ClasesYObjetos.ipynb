## Creando Clases

```python
class Car: # Molde

    def __init__(self, make, model, year): # Contructor  
        self.make = make
        self.model = model
        self.year = year
        print("Una instancia de la clase se ha creado exitosamente")

    def print_atributes(self):
        print("Marca:", self.make)
        print("Model:", self.model)
        print("Año:", self.year)
```

## Creando Objetos

```python
car1 = Car("aTesla", "Model 3", "2020")
```

```python
car1.print_atributes()
```

## Accediendo a los atributos

```python
car2 = Car("Tesla", "Model Y", "2019")
car2.print_atributes()
```

## Variables de Clase

```python
class Car: # Molde
    opus = 0

    def __init__(self, make, model, year): # Contructor  
        Car.opus += 1
        self.make = make
        self.model = model
        self.year = year
        print("Una instancia de la clase se ha creado exitosamente")

    def print_atributes(self):
        print("Marca:", self.make)
        print("Model:", self.model)
        print("Año:", self.year)
```

```python
car1 = Car("aTesla", "Model 3", "2020")
car2 = Car("Tesla", "Model Y", "2019")
```

## Garbage Collection

```python
class Car: # Molde
    total = 0

    def __init__(self, make, model, year): # Contructor  
        Car.total += 1
        self.make = make
        self.model = model
        self.year = year
        print("Una instancia de la clase se ha creado exitosamente")

    def __del__(self):
        Car.total -= 1

    def print_atributes(self):
        print("Marca:", self.make)
        print("Model:", self.model)
        print("Año:", self.year)
```

```python
car1 = Car("aTesla", "Model 3", "2020")
car2 = Car("Tesla", "Model Y", "2019")
car3 = Car("Tesla", "Model X", "2021")
car4 = Car("Tesla", "Model S", "2022")
```

```python
print(" hay", Car.total, "carros creados")
```

```python
del car4
```

```python
print(" hay", Car.total, "carros creados")
```

```python
car4.make
```

## Herencia

```python

class Car:
    super_atribute = "Atributo de la super clase / Clase Padre"
    def __init__(self):
        print("Inicialización de la súper clase / Clase Padre")
    def super_method(self):
        print("Metodo de la súper clase / Clase Padre")

class CollectionCar(Car): #  Hija que va a heredar
    sub_atribute = "Atributo de la sub clase / Clase Hija"
    def __init__(self):
        print("Inicialización de la sub clase / Clase Hija")
    def sub_method(self):
        print("Metodo de la sub clase / Clase Hija")
        
```

```python
sub_car = CollectionCar()
```

```python
sub_car.super_method()
sub_car.sub_method()
print(sub_car.super_atribute)
print(sub_car.sub_atribute)
```

## Usando una SubClase como SuperClase

```python

class Car:
    super_atribute = "Atributo de la super clase / Clase Padre / Abuelo"
    def __init__(self):
        print("Inicialización de la súper clase / Clase Padre / Abuelo")
    def super_method(self):
        print("Metodo de la súper clase / Clase Padre / Abuelo")

class CollectionCar(Car): #  Hija que va a heredar
    sub_atribute = "Atributo de la sub clase / Clase Hija"
    def __init__(self):
        print("Inicialización de la sub clase / Clase Hija")
    def sub_method(self):
        print("Metodo de la sub clase / Clase Hija")

class SubCollectionCar(CollectionCar): #  Nieta que va a heredar
    sub_sub_atribute = "Atributo de la sub clase / Clase Nieta"
    def __init__(self):
        print("Inicialización de la sub clase / Clase Nieta")
    def sub_sub_method(self):
        print("Metodo de la sub clase / Clase Nieta")
```

```python
sub_sub_car = SubCollectionCar()
```

```python
sub_sub_car.super_method()
sub_sub_car.sub_method()
sub_sub_car.sub_sub_method()
print(sub_sub_car.super_atribute)
print(sub_sub_car.sub_atribute)
print(sub_sub_car.sub_sub_atribute)
```
