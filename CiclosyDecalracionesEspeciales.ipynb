## Ciclos 
- Tambien son llamados Declaraciones Iterativas
- Son usados para repetir un bloque especifíco de código
- Si vas a hacer un proceso una y otra vez, entonces puedes usar un ciclo para ello

- Tipos
    - for
    - while

## Ciclo for
- Son usados para iterar los valores "n" número de veces
- Podemos usar la palabra reservada "for"
- Puede iterar los valores de estructuras de datos como listas, tuplas, sets o diccionarios
- También puede iterar sobre cada letra de una cadena
- Itera en una secuencia en orden mientras ejecuta un bloque de codigo

```python
my_string = "Backend Python"
print(my_string)
print(type(my_string))

for letter in my_string:
    print("antes")
    print(letter)
    print("despues")
```

```python
my_string = "Backend Python"
print(my_string)
print(type(my_string))

for character in my_string:
    print("Ciclo for iterando con la letra: ", character)
```

```python
my_list = [1,2,3,4,5]
print(my_list)
print(type(my_list))

for number in my_list:
    print("Ciclo for iterando con el numero: ", number)
```

```python
for number in range(1,6): # 6 - 1 = 5
    print("Ciclo for iterando con el numero: ", number)
```

```python
for number in range(1,5+1): # 6 - 1 = 5
    print("Ciclo for iterando con el numero: ", number)
```

## Números Pares e Impares con Ciclo For

```python
# Pares
for number in range(1,30+1): # 31 - 1 = 30
    #print("Ciclo for iterando con el numero: ", number)
    if number % 2 == 0:
        print(number, "es un numero par")
```

```python
# Impares
for number in range(1,30+1): # 31 - 1 = 30
    #print("Ciclo for iterando con el numero: ", number)
    if number % 2 == 1:
        print(number, "es un numero impar")
```

## Ciclo For anidado
- Significa un ciclo for dentro de otro ciclo for
- Es cuando tienes un ciclo dentro de otro ciclo
- Recuerda, usamos el ciclo for para hacer una tarea cierta cantidad de veces basado en una condición

- Ciclo principal
  - Ciclo anidado

```python
nouns = ["balon", "bote", "celular", "libro"]
adjetives = ["pequeño", "grande", "redondo", "cuadrado", "rojo", "azul"]

for noun in nouns:
    # 0 noun -> balon
    # 1 noun -> bote
    for adjetive in adjetives:
        # 0 adjetive -> pequeño
        print("{} {}".format(noun, adjetive))
```

## Ciclo While
- Solo se ejecuta si la condicion es verdadera
- Se puede usar para correr un proseso de manera perpetua
- Consiste de una condicion y un bloque de código
- Valida la condición antes de que el bloque sea ejecutado

```python
my_number = 5
while my_number < 5:
    print("el valor de my_number es: ", my_number)
```

```python
my_number = 5
print("previo al Ciclo while")
while my_number < 5:
    print("dentro del ciclo while")
    print("el valor de my_number es: ", my_number)
print("final de ejemplo")
```

```python
my_number = 0
print("previo al Ciclo while")
while my_number < 5:
    print("dentro del ciclo while")
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1
print("final de ejemplo")
```

```python
my_number = 0
print("previo al Ciclo while")
while my_number < 5:
    print("dentro del ciclo while")
    print("El valor de my_number es: ", my_number)
print("final de ejemplo")
```

```python
my_number = 0
print("previo al Ciclo while")
while my_number >= 0 and my_number < 5:
    print("dentro del ciclo while")
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1
else:
    print(my_number, " es un numero negativo o mayor a cuatro")
print("final de ejemplo")
```

```python
my_number = -3
print("previo al Ciclo while")
while my_number >= 0 and my_number < 5:
    print("dentro del ciclo while")
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1
else: # de lo contrario
    print(my_number, " es un numero negativo o mayor a cuatro")
print("final de ejemplo")
```

## Iterando en Estructuras de Datos

```python
# Listas
emails = ["carlos@test.com", "sofia@test.com", "alejandro@test.com"]

for email in emails:
    print(email)
```

```python
# Listas
user_ids = [1825, 2561, 3740]
           
for user_id in user_ids:
    print(user_id)
```

```python
user_ids = [1825, 2561, 3740]
emails = ["carlos@test.com", "sofia@test.com", "alejandro@test.com"]
my_dict = dict(zip(user_ids, emails))

print(my_dict)

for user_id, email in my_dict.items():
    print("el user {} tiene el correo {}".format(user_id, email))
```

```python
# Tuplas
user_ids = (1825, 2561, 3740)

print(user_ids)
print(type(user_ids))
           
for user_id in user_ids:
    print(user_id)
```

```python
# Sets
user_ids = {1825, 2561, 3740, 3740, 3740, 3740}

print(user_ids)
print(type(user_ids))
           
for user_id in user_ids:
    print(user_id)
```

## Declaraciones Especiales
- break: detiene el ciclo
- continue: se salta la iteración actual (sin ejecutar el código posterior) y pasar a la siguiente iteración
- pass: se usa para escribir ciclos vacíos o para escribir funciones o clases vacias

## break
- Es usado para detener un ciclo basado en ciertas condiciones
- Te ayuda a reducir el número de iteraciones innecesarias

```python
my_number = 0
past_number = None
print("previo al Ciclo while")
while my_number < 5:
    if past_number == my_number:
        break
    print("dentro del ciclo while")
    print("El valor de my_number es: ", my_number)
    past_number = my_number
print("final de ejemplo")
```

```python
my_number = 0
while my_number < 5:
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1 
print("final de ejemplo")
```

```python
my_number = 0
while my_number < 5:
    if my_number == 3:
        print("antes del break")
        break
        print("despues del break")
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1 
print("final de ejemplo")
```

## continue
- Salta la iteración actual y pasa a la siguiente iteración
- Se puede usar tanto en ciclos for como en ciclos while

```python
my_number = 0
while my_number < 5:
    if my_number == 3:
        my_number += 1
        print("antes del continue")
        continue
        print("despues del continue")
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1 
print("final de ejemplo")
```

```python
for number in range(1,20 + 1): # 21 - 1 = 20
    if number == 10:
        continue
    print("Ciclo for iterando con le número: ", number) 
```
