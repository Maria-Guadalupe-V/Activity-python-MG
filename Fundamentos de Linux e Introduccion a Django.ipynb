```python
#views.py
import os

from django import get_version
from django.conf import settings
from django.shortcuts import render


class User():
    # atributos
    def __init__(self, name, email, password):
        self.name = name
        self.email = email
        self.password = password
    # metodos
    def login(self):
        print( self.name, "inicio sesion")
    def logout(self):
        print(self.name, "cerro sesion")
    def viewProfile(self):
        print("Perfil")
        print("Nombre: ", self.name)
        print("Correo: ", self.email)

class Customer(User):
    #atributos
    def __init__(self, nombre, correo, contraseña, address):
        super().__init__(nombre, correo, contraseña)
        self.address = address
        self.carrito = 0
    # metodos
    def addToCar(self):
        self.carrito += 1
        print("objeto agregado al carrito")
    def removeFromTheCar(self):
        if self.carrito > 0:
            self.carrito -= 1
            print("objeto removido de el carrito")
        else:
            print("el carrito esta vacio")
    def cartItems(self):
        print("Hay", self.carrito, "objetos en el carro")
    def changeAddress(self, newAddress):
        self.address = newAddress
        print("Direccion actualizada")

usuario = User("Maria", "maria@gmail.com", "1234")
cliente = Customer("Lupe", "lupe@gmail.com", "5678", "Michoacan")

def home(request):
    context = {
        "debug": settings.DEBUG,
        "django_ver": get_version() + "PROBANDO CAMBIOS",
        "python_ver": os.environ["PYTHON_VERSION"] + "CAMBIOS",
        "usuario": usuario.email,
        "cliente": cliente.email,
    }

    return render(request, "pages/home.html", context)

```
