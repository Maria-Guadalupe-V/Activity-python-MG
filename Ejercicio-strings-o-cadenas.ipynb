```python
while True:
    print(" Analizador de sentimientos (escribe 'salir' para terminar)\n")
    my_string = input("Ingresa el texto:\n").lower()
    if my_string.strip() == "salir":
        print("👋 Programa finalizado.")
        break
    positive_words = [
        "bueno", "excelente", "maravilloso", "fantástico", "genial", "agradable", "feliz", "increíble",
        "espectacular", "hermoso", "impresionante", "positivo", "brillante", "alegre", "satisfactorio",
        "perfecto", "excepcional", "útil", "amable", "fácil", "ameno", "rico", "sabroso", "eficiente",
        "valioso", "productivo", "respetuoso", "divertido", "asombroso", "ideal", "eficaz"
       ]

    negative_words = [
        "malo", "horrible", "terrible", "pésimo", "defectuoso", "feo", "desagradable", "triste", "aburrido",
        "decepcionante", "frustrante", "difícil", "complicado", "lento", "inútil", "estresante",
        "negativo", "molesto", "problemático", "fallido", "mediocre", "desastroso", "ridículo", "malísimo",
        "engorroso", "asqueroso", "inaceptable", "inservible", "insatisfactorio", "basura"
        ]

    score = 0

    for word in positive_words:
        score += my_string.count(word)

    for word in negative_words:
        score -= my_string.count(word)

    if score > 0:
        result = "Positivo"
    elif score < 0:
        result = "Negativo"
    else:
        result = "Neutro"

    print("\nAnalizador de sentimientos")
    print("Texto:", my_string)
    print("Puntaje:", score)
    print("Resultado:", result)
```
