## Diferentes formas de crear diccionarios

```python
my_dict = {}
print(my_dict)
print(type(my_dict))
```

```python
my_dict = dict()
print(my_dict)
print(type(my_dict))
```

```python
# representando el diccionario como tal
my_dict = {
    1:"Uno",
    2:"Dos",
    3:"Tres",
    4:"Cuatro",
    5:"Cinco"
}
print(my_dict)
print(type(my_dict))
```

```python
# usando nombres de las llaves con valores
my_dict = dict(Uno=1, Dos=2, Tres=3, Cuatro=4, Cinco=5)
print(my_dict)
print(type(my_dict))
```

```python
# usando una lista de tuplas
my_dict = dict([
    (1, "Uno"),
    (2, "Dos"),
    (3, "Tres"),
    (4, "Cuatro"),
    (5, "Cinco"),
])
print(my_dict)
print(type(my_dict))
```

## Método update

```python
my_dict = {
    1:"Uno",
    2:"Dos",
    3:"Tres",
    4:"Cuatro",
    5:"Cinco"
}
my_dict_2 = {
    6:"Seis",
    7:"Siete",
    8:"Ocho",
    9:"Nueve"
    
}
print(my_dict)
print(type(my_dict))

print(my_dict_2)
print(type(my_dict_2))
```

```python
my_dict.update(my_dict_2)
print(my_dict)
print(type(my_dict))
```

```python
my_dict = {
    1:"Uno",
    2:"Dos",
    3:"Tres",
    4:"Cuatro",
    5:"Cinco"
}
my_dict_2 = {
    1:"UNO",
    3:"TRES",
    4:"CUATRO",
}
print(my_dict)
print(type(my_dict))

print(my_dict_2)
print(type(my_dict_2))

my_dict.update(my_dict_2)
print(my_dict)
print(type(my_dict))
```

## Copy & Alias

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}
# usar un alias va a afectar al dicciinario original
alias = laptops # no es recomendable
print(alias)
alias["HP"] = 9999.0
print(alias)
print(laptops)
```

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}
# usar copy es la forma correcta, ya que mantiene los valores del diccionario original
copy = laptops.copy() # siempre es recomendable
print(copy)
copy["HP"] = 9999.0
print(copy)
print(laptops)
```

## Obteniendo las keys/llaves y values/valores por separado

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}
```

### keys

```python
print(laptops.keys())
```

```python
for key in laptops.keys():
    print("Para la llave '{}' el valor es '{}'".format(key, laptops[key]))
```

### values

```python
print(laptops.values())
```

```python
for value in laptops.values():
    print("El valor es {}".format(value))
```

```python
def get_key(d, val):
    keys = [k for k, v in d.items() if v == val]
    if keys:
        return keys[0]
    return None 

for value in laptops.values():
    print("Para la llave '{}' el valor es '{}'".format(get_key(laptops, value), value))
```

## Métodos para manipulación de Diccionarios

### clear
eliminar los elementos de un diccionario

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

print(laptops)
print(type(laptops))
print(len(laptops))

laptops.clear()
print(laptops)
print(type(laptops))
print(len(laptops))
```

### copy
Hacer una copia del diccionario

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

print(laptops)
print(type(laptops))
print(len(laptops))

copy = laptops.copy()

print(copy)
print(type(copy))
print(len(copy))

copy.clear()

print(laptops)
print(type(laptops))
print(len(laptops))

print(copy)
print(type(copy))
print(len(copy))
```

### fromkeys
contruir un nuevo diccionario tomando como referencia las keys/llaves de un diccionario

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

category = "Marca de laptop"
new_dict = dict.fromkeys(laptops, category)
print(new_dict)
```

### get
obtener el valor de  un diccionario de forma segura y con la opción de configurar un valor por default

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

print(laptops["Lenovo"])
print(laptops.get("Lenovo"))
```

```python
# forma insegura de obtener un valor 
print(laptops["Asus"])
```

```python
# forma segura de obtener un valor 
print(laptops.get("Asus"))
```

```python
# forma segura de obtener un valor y configurando valor por default
print(laptops.get("Asus", 0))
print(laptops.get("Asus", "No disponible"))
```

```python
# forma segura de obtener un valor y configurando valor por default
print(laptops.get("Lenovo", 0))
print(laptops.get("Lenovo", "No disponible"))
```

### pop 
elimina la llave y valor de un diccionario pasando la llave como parámetro

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

print(laptops)
print(len(laptops))

laptops.pop("Lenovo")
print(laptops)
print(len(laptops))
```

```python
laptops.pop("Asus")
print(laptops)
print(len(laptops))
```

```python
print(laptops)
print(len(laptops))

if "Asus" in laptops.keys():
    laptops.pop("Asus")
    print(laptops)
    print(len(laptops))
else: 
    print("La llave Asus no esta en el diccionario")
```

### popitem
Elimina el indice -1 (El último elemento llave-valor)

```python
laptops = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

print(laptops)
print(len(laptops))

laptops.popitem() # el indice -1 (el último elemento llave-valor)
print(laptops)
print(len(laptops))

laptops.popitem() # el indice -1 (el último elemento llave-valor)
print(laptops)
print(len(laptops))

laptops.popitem() # el indice -1 (el último elemento llave-valor)
print(laptops)
print(len(laptops))
```

### setdefault
Configura un valor por default para cierta llave en caso de que no éste presente en el diccionario.

Sí la llave está presente, no tendrá ningún efecto. Pero si la llave no está, entonces configurará el valor para esa llave

```python
laptops_store_1 = {
    "HP": 11999.0,
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

laptops_store_2 = {
    "Apple": 24999.0,
    "Acer": 12012.03,
    "Lenovo": 17299.0,
    "Dell": 15524.87
}

laptops_store_1.setdefault("HP", 10000.0)
laptops_store_2.setdefault("HP", 10000.0)

print(laptops_store_1.get("HP"))
print(laptops_store_2.get("HP"))
```
