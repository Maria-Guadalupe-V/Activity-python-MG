# Calculadora

```python
 while True:
    print("===CALCULADORA===")
    print("Elige una opcion")
    print("1-suma")
    print("2-resta")
    print("3-multiplicacion")
    print("4-divicion")
    print("5-potencia")
    print("6-salir")
    x= input("Escribe tu eleccion")
    y= float(x)
    
    if(y==1):
         a= input("Ingresa un numero\n")
         b= input("Ingresa otro numero\n")
         a2= float(a)
         b2= float(b)
         resultado= a2 + b2
         print("El resultado es:" ,resultado )
    elif(y==2):
        a= input("Ingresa un numero\n")
        b= input("Ingresa otro numero\n")
        a2= float(a)
        b2= float(b)
        resultado= a2 - b2
        print("El resultado es:" ,resultado )
    elif(y==3):
        a= input("Ingresa un numero\n")
        b= input("Ingresa otro numero\n")
        a2= float(a)
        b2= float(b)
        resultado= a2 * b2
        print("El resultado es:" ,resultado )
    elif(y==4):
        a= input("Ingresa un numero\n")
        b= input("Ingresa otro numero\n")
        a2= float(a)
        b2= float(b)
        resultado= a2/b2
        print("El resultado es:" ,resultado)
    elif(y==5):
        a= input("Ingresa un numero\n")
        b= input("Ingresa otro numero\n")
        a2= float(a)
        b2= float(b)
        resultado= a2**b2
        print("El resultado es:" ,resultado)
    else:
        break
```
