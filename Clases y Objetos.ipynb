```python
class User():
    # atributos
    def __init__(self, name, email, password):
        self.name = name
        self.email = email
        self.password = password
    # metodos
    def login(self):
        print( self.name, "inicio sesion")
    def logout(self):
        print(self.name, "cerro sesion")
    def viewProfile(self):
        print("Perfil")
        print("Nombre: ", self.name)
        print("Correo: ", self.email)
```

```python
usuario1 = User(input("escribe tu nombre"), input("escribe tu correo"), input("escribe tu contraseña"))
```

```python
usuario1.login()
```

```python
usuario1.logout()
```

```python
usuario1.viewProfile()
```

```python
class Customer(User):
    #atributos
    def __init__(self, nombre, correo, contraseña, address):
        super().__init__(nombre, correo, contraseña)
        self.address = address
        self.carrito = 0
    # metodos
    def addToCar(self):
        self.carrito += 1
        print("objeto agregado al carrito")
    def removeFromTheCar(self):
        if self.carrito > 0:
            self.carrito -= 1
            print("objeto removido de el carrito")
        else:
            print("el carrito esta vacio")
    def cartItems(self):
        print("Hay", self.carrito, "objetos en el carro")
    def changeAddress(self, newAddress):
        self.address = newAddress
        print("Direccion actualizada")
```

```python
carrito1 = Customer(input("escribe tu nombre"), input("escribe tu correo"), input("escribe tu contraseña"),input("ingresa tu direccion"))
```

```python
carrito1.login()
```

```python
carrito1.logout()
```

```python
carrito1.viewProfile()
```

```python
carrito1.addToCar()
```

```python
carrito1.removeFromTheCar()
```

```python
carrito1.removeFromTheCar()
```

```python
carrito1.cartItems()
```
